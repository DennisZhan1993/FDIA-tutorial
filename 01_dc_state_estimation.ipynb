## Measurement model

We use

\[
z = Hx + e,
\]

where bus 1 is the angle reference and

\[
x=[\theta_2,\theta_3]^T.
\]


```python
from src.grid_model import build_three_bus_model, generate_measurement
from src.state_estimator import wls_estimate

rng = np.random.default_rng(42)
model = build_three_bus_model(sigma=0.01)

print("H =\n", model.H)
print("x_true =", model.x_true)
print("Measurements =", model.measurement_names)
```

```python
z, noise = generate_measurement(model, rng)
result = wls_estimate(z, model.H, model.R)

print("z =", z)
print("noise =", noise)
print("x_hat =", result.x_hat)
print("estimation error =", result.x_hat - model.x_true)
print("residual =", result.residual)
print("J =", result.J)
```
